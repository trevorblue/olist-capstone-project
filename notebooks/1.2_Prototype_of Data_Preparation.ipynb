```python
import pandas as pd
import  os

DATA_DIR =os.path.join("..", "data", "raw")  # Define the data directory

summary = []    # Create an empty list to store results

for file in os.listdir(DATA_DIR): # Loop through all files in the folder
    if file.endswith(".csv"): # Process only CSV files# Filter only CSV files
        path = os.path.join(DATA_DIR, file) # Get the full file path# Build the full file path
        df = pd.read_csv(path)      # . Read the CSV into a DataFrame

        summary.append({        # Collect metadata (not the data)
            "dataset": file,
            "rows": df.shape[0],
            "columns": list(df.columns)
        })

# Display the results
for item in summary:
    print(f"\nDataset: {item['dataset']}")
    print(f"Rows: {item['rows']}")
    print("Columns:")
    for col in item['columns']:
            print(f" - {col}")
```

```python
import pandas as pd
import os

DATA_DIR = os.path.join("..", "data", "raw")

# First, let's check if the directory exists and what's in it
print(f"Checking directory: {DATA_DIR}")
print(f"Directory exists: {os.path.exists(DATA_DIR)}")
print(f"Is it a directory: {os.path.isdir(DATA_DIR)}")

# List ALL files in the directory
print("\nAll files in directory:")
try:
    all_files = os.listdir(DATA_DIR)
    print(f"Total files found: {len(all_files)}")
    for file in all_files:
        print(f"  - {file}")
except Exception as e:
    print(f"Error: {e}")

# Now try to process CSV files
print("\n" + "="*50)
print("Processing CSV files:")
print("="*50)

for file in os.listdir(DATA_DIR):
    print(f"Checking file: {file}")  # Debug line
    if file.endswith(".csv"):
        print(f" Found CSV: {file}")  # Debug line
        path = os.path.join(DATA_DIR, file)
        df = pd.read_csv(path)
        print(f"\n {file}")
        print(df.dtypes)
    else:
        print(f"Skipped: {file}")  # Debug line
```

```python
# Detecting possible JOIN keys between datasets

from collections import defaultdict

'''Imports defaultdict from Python’s collections module

defaultdict is a special dictionary that:

Automatically creates a default value when a key does not exist

In this case, the default value will be an empty list

Why this matters:

Avoids manually checking if a key already exists

Makes the code cleaner and safer'''
 
column_map = defaultdict(list)

for file in os.listdir(DATA_DIR):
    if file.endswith(".csv"):
        df = pd.read_csv(os.path.join(DATA_DIR, file))
        for col in df.columns:
            column_map[col].append(file)

# Display potential join keys
for col, files in column_map.items():
    if len(files) > 1:
        print(f"Column '{col}' found in files: {files}")
```
